## 반복 점검 함수

````python
def inspect_dataframe(

    name: str,

    df: pd.DataFrame,

    primary_key: str | None = None,

) -> dict:

    result = {

        "dataset": name,

        "rows": df.shape[0],

        "columns": df.shape[1],

        "column_names": df.columns.tolist(),

        "dtypes": {

            column: str(dtype)

            for column, dtype in df.dtypes.items()

        },

        "missing_total": int(df.isna().sum().sum()),

        "duplicate_rows": int(df.duplicated().sum()),

    }

    if primary_key is not None:

        result["primary_key"] = primary_key

        result["key_missing"] = int(

            df[primary_key].isna().sum()

        )

        result["key_duplicates"] = int(

            df[primary_key].duplicated().sum()

        )

        result["key_unique"] = int(

            df[primary_key].nunique()

        )

    return result

 
````

````python
inspection_results = []

for name, df in datasets.items():

    inspection_results.append(

        inspect_dataframe(

            name,

            df,

            primary_keys[name],

        )

    )

inspection_results
````

# 데이터 구조 및 품질 점검 요약 LLM보고서

## 1. 데이터셋 기본 구조 및 식별자
분석 시 각 데이터가 어떤 단위를 가지며, 서로 어떻게 연결되는지 정의합니다.

| 데이터셋 | 1행(Row)의 의미 | 주요 ID (Primary Key) | 연결 고리 (Foreign Key) |
|:---|:---|:---|:---|
| **`customers`** | 쇼핑몰에 가입한 고객 1명의 고유 프로필 | `customer_id` | 해당 없음 |
| **`products`** | 쇼핑몰에서 판매 중인 독립 상품 1개 | `product_id` | 해당 없음 |
| **`orders`** | 고객이 결제를 완료한 주문(장바구니) 1건 | `order_id` | `customer_id` (`customers` 참조) |
| **`order_items`** | 주문 내역(영수증) 안의 개별 품목 1줄 | `order_item_id` | `order_id` (`orders` 참조)<br>`product_id` (`products` 참조) |

<br>

## 2. 삭제 금지 및 주의 데이터

구조 점검 요약 기준으로 `missing_total`, `duplicate_rows`, `key_duplicates`, FK `unknown_count`가 **모든 데이터셋에서 0건**으로 확인되어, 현재 시점에는 즉시 삭제를 검토해야 할 결측·중복·고아 레코드가 없습니다.

다만 향후 데이터가 갱신되거나 실제 원본을 직접 열람할 경우, 아래 유형이 발견되면 **건수와 상관없이 무작정 삭제하지 말고 원인부터 확인**해야 합니다.

| 데이터셋 | 문제 종류 | 현재 발생 건수 | 즉시 삭제하면 안 되는 이유 (위험성) |
|:---|:---|---:|:---|
| **`orders`** | 참조 불가 고객 (customers에 없는 customer_id) | 0건 | 탈퇴 회원의 과거 주문일 수 있음. 삭제 시 전체 주문·매출 통계 왜곡 |
| **`order_items`** | 참조 불가 주문 (orders에 없는 order_id) | 0건 | 시스템 지연 등으로 매칭이 꼬였을 수 있음. 삭제 시 실제 판매된 품목 소실 |
| **`order_items`** | 참조 불가 상품 (products에 없는 product_id) | 0건 | 단종되어 상품 마스터에서 제외된 항목일 수 있음. 삭제 시 과거 판매 기록 소실 |
| **`order_items`** | 주요 ID(`order_item_id`) 중복 | 0건 | 전산 오류일 수도, 동일 상품의 별도 라인 항목일 수도 있음. 확인 없이 삭제 시 매출 누락 위험 |

<br>

## 3. 추가 점검 우선순위 및 Pandas 코드 목적
요약본에는 드러나지 않지만, 실제 분석을 시작하기 전 반드시 코드로 재확인해야 할 잠재적 위험 요소입니다.

| 순위 | 점검 대상 컬럼 | 예상되는 구조적 문제 | Pandas 코드 실행 목적 (검증 방법) |
|:---:|:---|:---|:---|
| **1순위** | `signup_date`<br>`order_date` | 날짜 데이터가 문자열(`str`)로 기록되어 있어 기간 계산·정렬이 불가능함 | 문자열을 `datetime`으로 강제 변환하여, 형식이 어긋나 깨지는(NaT) 불량 데이터 색출 |
| **2순위** | `age`, `price`<br>`quantity`, `unit_price` | 정수형(`int64`)이나, 음수 나이 또는 0원 이하 가격 등 상식 밖 수치가 있을 수 있음 | 기초 통계량(최소/최대/평균)을 추출하여 현실적으로 불가능한 극단치(이상치) 감지 |
| **3순위** | 품목 없는 주문 여부 | `orders`에는 존재하나 `order_items`에 매칭 행이 하나도 없는 주문이 있을 수 있음 | `orders.order_id`가 `order_items.order_id`에 전부 존재하는지 대조하여 매출 집계 누락 여부 확인 |
